```python
import numpy as np

def calf_len(df):
    n = len(df)
    core = df.iloc[int(0.1 * n): int(0.9 * n)]      # drop the first and last 10% of frames

    dx = core['right_ankle_x'] - core['right_knee_x']
    dy = core['right_ankle_y'] - core['right_knee_y']
    calf_len = np.hypot(dx, dy).median() 
    return calf_len.round(2)

calflen = calf_len(df)
calflen
```

```python
import numpy as np

window = 5
smooth_L = df['rel_ankle_left'].rolling(window, center=True, min_periods=1).median()
smooth_R = df['rel_ankle_right'].rolling(window, center=True, min_periods=1).median()

dev_L = df['rel_ankle_left'] - smooth_L     # how far each frame is from its "expected" value
dev_R = df['rel_ankle_right'] - smooth_R

threshold = 0.5 * calflen                # tune this (see below)
df['swap'] = ((dev_L.abs() > threshold) &
              (dev_R.abs() > threshold) &
              (np.sign(dev_L) != np.sign(dev_R)))

df
```

```python
import numpy as np
from scipy.signal import find_peaks

sig = df['rel_ankle_right_new'].to_numpy()
prom = 0.3 * (sig.max() - sig.min())          # a real peak must stand out this much

peaks, _   = find_peaks(sig,  prominence=prom)   # foot furthest in front
troughs, _ = find_peaks(-sig, prominence=prom)   # foot furthest behind

gaps = np.concatenate([np.diff(peaks), np.diff(troughs)])
stride_frames = np.median(gaps)

print("peak frames:  ", df['frame'].iloc[peaks].values)
print("trough frames:", df['frame'].iloc[troughs].values)
print("stride period:", stride_frames, "frames")
print(sig[peaks])
```

```python
from scipy.signal import butter, filtfilt

k = 6                              # keep movement up to 6x the stride frequency
Wn = (k / stride_frames) / 0.5         # as a fraction of the Nyquist limit (0.5 cycles/frame)
b, a = butter(4, Wn, btype='low')

df_s = df.copy()                       # keep the unsmoothed data for comparison
for c in x_cols + y_cols:
    df_s[c] = filtfilt(b, a, df[c].to_numpy())

df_s['mid_hip_x'] = (df_s['left_hip_x'] + df_s['right_hip_x']) / 2
df_s['rel_ankle_left_new'] = df_s['left_ankle_x'] - df_s['mid_hip_x']
df_s['rel_ankle_right_new'] = df_s['right_ankle_x'] - df_s['mid_hip_x']

cols = ['mid_hip_x', 'rel_ankle_right_new']   # your 3 columns

fig, axes = plt.subplots(1, 2, figsize=(18, 4), sharex=True)
for ax, col in zip(axes, cols):
    ax.plot(df['frame'], df_s[col])
    ax.set_title(col)
    ax.set_xlabel('frame')

plt.tight_layout()
plt.show()
```

```python
def facing_right(df):
    foot_ankle_diff_left = (df['left_foot_index_x'] - df['left_heel_x']).median()
    foot_ankle_diff_right = (df['right_foot_index_x'] - df['right_heel_x']).median()
    if (foot_ankle_diff_left > 0) and (foot_ankle_diff_right > 0):
        return 1
    elif (foot_ankle_diff_left < 0) and (foot_ankle_diff_right < 0):
        return -1
    else:
        return 0
    
def body_near(df):
    if facing_right(df) == 1:
        return 'right'
    elif facing_right(df) == -1:
        return 'left'
    else:
        return 'NA'
    
body_near(df)
```

```python
# write the function to detect touchdown and midstance frame

import numpy as np
from scipy.signal import argrelextrema

# the frame that foot steps on floor
def detect_touchdowns(df):
    f = f'rel_ankle_{body_near(df)}_new'
    peaks = argrelextrema(np.array(df[f]), np.greater)
    return [df.loc[p, 'frame'] for p in peaks][0].tolist()

def peaks_error(peaks):
    error_idx = []
    for i in range(len(peaks) - 1):
        gap = peaks[i+1] - peaks[i]
        if gap < 0.5 * stride_frames:                         # fixed: only flag gaps much shorter than a stride
            error_idx.append(i)

    if len(error_idx) == 0:
        return []                                             # fixed: empty list means no errors
    else:
        return error_idx

# the frame that mid-hip pass the foot that stepping
def midstance(df):
    f = f'rel_ankle_{body_near(df)}_new'
    rel_ankle = np.array(df[f])                               # fixed: renamed (not always left)
    sign = np.sign(rel_ankle)
    mask = (sign[:-1] > 0) & (sign[1:] <= 0)                  # fixed: only + to - crossings
    idx = np.where(mask)
    return [df.loc[p, 'frame'] for p in idx][0].tolist()

# td =  detect_touchdowns(df_s)
# peaks_error(td)
midstance(df_s)
```

```python
# write the function to return angle
import math
import numpy as np
def joint_angle(a: list, b: list, c: list):
    u = [a[0] - b[0], a[1] - b[1]]
    v = [c[0] - b[0], c[1] - b[1]]
    u_dot_v = (u[0] * v[0]) + (u[1] * v[1])
    u_norm = math.sqrt((u[0])**2 + (u[1])**2)
    v_norm = math.sqrt((v[0])**2 + (v[1])**2)
    cos_theta = u_dot_v / (u_norm * v_norm)
    theta_rad = np.arccos(np.clip(cos_theta, -1, 1))
    theta_deg = np.degrees(theta_rad)
    return theta_deg.round(2)

# compute the knee reflexion for all frame
hip_x = f'{body_near(df_s)}_hip_x'
hip_y = f'{body_near(df_s)}_hip_y'
knee_x = f'{body_near(df_s)}_knee_x'
knee_y = f'{body_near(df_s)}_knee_y'
ankle_x = f'{body_near(df_s)}_ankle_x'
ankle_y = f'{body_near(df_s)}_ankle_y'

df_s['knee_angle'] = df_s.apply(
    lambda r: joint_angle([r[hip_x], r[hip_y]],
                          [r[knee_x], r[knee_y]],
                          [r[ankle_x], r[ankle_y]]),
    axis=1
)

df_s['knee_flexion'] = (180 - df_s['knee_angle']).round(2)

df_s
```

```python
# compute shank angle
def shank_angle(df):
    df1 = df.copy()
    bodynear = body_near(df1)
    ax = f'{bodynear}_ankle_x'
    ay = f'{bodynear}_ankle_y'
    kx = f'{bodynear}_knee_x'
    ky = f'{bodynear}_knee_y'
    df1['dx'] = (df1[ax] - df1[kx]) * facing_right(df1)
    df1['dy'] = df1[ky] - df1[ay]
    df1['theta_deg'] = (np.degrees(np.arctan2(df1['dx'], df1['dy']))).round(2)
    return df1['theta_deg']

df_s['shank_angle'] = shank_angle(df_s)
df_s
```

```python
# compute trunk lean angle
def trunklean_angle(df):
    df1 = df.copy()
    sx = 'mid_shoulder_x'
    sy = 'mid_shoulder_y'
    hx = 'mid_hip_x'
    hy = 'mid_hip_y'
    df1['dx'] = (df1[sx] - df1[hx]) * facing_right(df1)
    df1['dy'] = df1[sy] - df1[hy]
    df1['theta_deg'] = (np.degrees(np.arctan2(df1['dx'], df1['dy']))).round(2)
    return df1['theta_deg']

df_s['trunklean_angle'] = trunklean_angle(df_s)
df_s[(df_s['frame'] >= 62) & (df_s['frame'] < 144)]['trunklean_angle'].mean()
```

```python
# compute arm swing angle
def armswing_angle(df):
    df1 = df.copy()
    bodynear = body_near(df)
    sx = f'{bodynear}_shoulder_x'
    sy = f'{bodynear}_shoulder_y'
    ex = f'{bodynear}_elbow_x'
    ey = f'{bodynear}_elbow_y'
    df1['dx'] = (df1[ex] - df1[sx]) * facing_right(df1)
    df1['dy'] = df1[sy] - df1[ey]
    df1['theta_deg'] = (np.degrees(np.arctan2(df1['dx'], df1['dy']))).round(2)
    return df1['theta_deg']

df_s['armswing_angle'] = armswing_angle(df_s)
df_s
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt


def to_img(x, y, H):
    """Convert our coordinates (y up) back to OpenCV pixels (y down)."""
    return int(round(x)), int(round(H - y))


def draw_check(video_path, df, frame_no, near, title, show_shank=False):
    # read the video frame
    cap = cv2.VideoCapture(video_path)
    cap.set(cv2.CAP_PROP_POS_FRAMES, int(frame_no))
    ok, img = cap.read()
    cap.release()
    if not ok:
        raise ValueError(f"could not read frame {frame_no}")
    H = img.shape[0]

    # the row for this frame (look up by the 'frame' column)
    r = df[df['frame'] == frame_no].iloc[0]

    def pt(name):  # near-side landmark, e.g. pt('knee')
        return to_img(r[f'{near}_{name}_x'], r[f'{near}_{name}_y'], H)

    sho, elb, wri = pt('shoulder'), pt('elbow'), pt('wrist')
    hip, knee, ank = pt('hip'), pt('knee'), pt('ankle')
    mhip = to_img(r['mid_hip_x'], r['mid_hip_y'], H)
    msho = to_img(r['mid_shoulder_x'], r['mid_shoulder_y'], H)

    # skeleton: arm and leg in green, trunk in blue, joints in red (OpenCV colours are BGR)
    for a, b in [(sho, elb), (elb, wri), (hip, knee), (knee, ank)]:
        cv2.line(img, a, b, (0, 255, 0), 2)
    cv2.line(img, mhip, msho, (255, 0, 0), 2)
    for p in [sho, elb, wri, hip, knee, ank]:
        cv2.circle(img, p, 4, (0, 0, 255), -1)

    # vertical reference line from the knee down to ankle height (yellow)
    if show_shank:
        cv2.line(img, knee, (knee[0], ank[1]), (0, 255, 255), 1)

    # metric values (Hershey fonts have no degree symbol, so none is used)
    lines = [title,
             f"knee flexion {r['knee_flexion']:.1f}",
             f"shank angle  {r['shank_angle']:.1f}",
             f"trunk lean   {r['trunklean_angle']:.1f}",
             f"arm swing    {r['armswing_angle']:.1f}"]
    for i, t in enumerate(lines):
        cv2.putText(img, t, (10, 20 + 18 * i), cv2.FONT_HERSHEY_SIMPLEX,
                    0.5, (0, 0, 0), 1, cv2.LINE_AA)
    return img


# ---- usage ----
near = body_near(df_s)
events = ([(f, 'touchdown', True) for f in td_frame] +
          [(f, 'midstance', False) for f in ms_frame])

fig, axes = plt.subplots(1, len(events), figsize=(5 * len(events), 4))
for ax, (f, name, shank) in zip(np.atleast_1d(axes), events):
    img = draw_check('run1.mp4', df_s, f, near, f'{name} frame {f}', show_shank=shank)
    ax.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))   # OpenCV is BGR, matplotlib expects RGB
    ax.set_axis_off()
plt.tight_layout()
plt.show()
```
