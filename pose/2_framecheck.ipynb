```python
import cv2

video_path = "run1.mp4"
frame_number = 144  # frames start at 0

cap = cv2.VideoCapture(video_path)
total = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
print(f"Total frames: {total}")

cap.set(cv2.CAP_PROP_POS_FRAMES, frame_number)  # jump to the frame
ok, frame = cap.read()
cap.release()

if ok:
    cv2.imshow(f"Frame {frame_number}", frame)
    cv2.waitKey(0)            # press any key to close
    cv2.destroyAllWindows()
    cv2.imwrite(f"frame_{frame_number}.png", frame)  # optional: save it
else:
    print("Could not read that frame")
```
