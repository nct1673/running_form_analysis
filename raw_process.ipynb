# **OpenWeather API**

```python
from datetime import datetime
import pytz

def myt2unix(tt):
    malaysia_tz = pytz.timezone('Asia/Kuala_Lumpur')

    # Parse the string to datetime, then localize to Malaysia time
    dt = datetime.strptime(tt, '%Y-%m-%d %H:%M:%S')
    dt_malaysia = malaysia_tz.localize(dt)

    # Convert to Unix timestamp
    unix_time = int(dt_malaysia.timestamp())
    return unix_time

df['dt_unix'] = df['api_start_time_local'].apply(myt2unix)
df
```

```python
import time

df = df.reset_index(drop=True)

yes = 0
results = []
call_count = 0  # ✅ track API calls

for i in range(len(df)):
    if pd.notna(df.loc[i, 'api_start_latitude']):
        tt_unix = df.loc[i, 'dt_unix']
        lat = round(df.loc[i, 'api_start_latitude'], 4)
        lon = round(df.loc[i, 'api_start_longitude'], 4)
        API_KEY = "867473ff72aa87abf0739682654da3ce"

        url = "https://api.openweathermap.org/data/4.0/onecall/timeline/1h?"
        params = {
            "appid": API_KEY,
            "lat": lat,
            "lon": lon,
            "dt": tt_unix
        }

        # ✅ Pause every 50 calls
        if call_count > 0 and call_count % 50 == 0:
            print(f"Reached {call_count} calls — waiting 60 seconds...")
            time.sleep(60)
            print("Resuming...")

        try:
            response = requests.get(url, params=params, timeout=10)
            if response.status_code == 200:
                data = response.json()
                hourly_list = data.get('data', [])
                
                # ✅ Find the entry where dt matches tt_unix (or closest one)
                matched = None
                for entry in hourly_list:
                    if entry['dt'] == tt_unix:
                        matched = entry
                        break
                
                # If no exact match, pick the closest dt
                if matched is None and hourly_list:
                    matched = min(hourly_list, key=lambda x: abs(x['dt'] - tt_unix))

                if matched:
                    results.append({
                        'index': i,
                        'weather_dt': matched.get('dt'),
                        'weather_temp': matched.get('temp'),
                        'weather_feels_like': matched.get('feels_like'),
                        'weather_pressure': matched.get('pressure'),
                        'weather_humidity': matched.get('humidity'),
                        'weather_dew_point': matched.get('dew_point'),
                        'weather_uvi': matched.get('uvi'),
                        'weather_clouds': matched.get('clouds'),
                        'weather_visibility': matched.get('visibility'),
                        'weather_wind_speed': matched.get('wind_speed'),
                        'weather_wind_deg': matched.get('wind_deg'),
                        'weather_wind_gust': matched.get('wind_gust'),          # may be None if absent
                        'weather_main': matched.get('weather', [{}])[0].get('main'),
                        'weather_description': matched.get('weather', [{}])[0].get('description'),
                        'weather_rain_1h': matched.get('rain', {}).get('1h'),   # None if no rain
                    })
                    yes += 1
                else:
                    results.append({'index': i, 'weather_dt': None, 'weather_temp': None,
                                    'weather_feels_like': None, 'weather_pressure': None,
                                    'weather_humidity': None, 'weather_dew_point': None,
                                    'weather_uvi': None, 'weather_clouds': None,
                                    'weather_visibility': None, 'weather_wind_speed': None,
                                    'weather_wind_deg': None, 'weather_wind_gust': None,
                                    'weather_main': None, 'weather_description': None,
                                    'weather_rain_1h': None})
            # if response.status_code == 200:
            #     data = response.json()
            #     hourly = data.get('data', [{}])[0]
            #     results.append({
            #         'index': i,
            #         'weather_temp': hourly.get('temp'),
            #         'weather_humidity': hourly.get('humidity'),
            #         'weather_description': hourly.get('weather', [{}])[0].get('description'),
            #     })
            #     yes += 1
            # else:
            #     print(f"Row {i}: API error {response.status_code}")
            #     results.append({'index': i, 'weather_temp': None, 'weather_humidity': None, 'weather_description': None})

        except requests.exceptions.ConnectTimeout:
            print(f"Row {i}: Timed out, skipping...")
            results.append({'index': i, 'weather_temp': None, 'weather_humidity': None, 'weather_description': None})

        except requests.exceptions.ConnectionError:
            print(f"Row {i}: Connection error, skipping...")
            results.append({'index': i, 'weather_temp': None, 'weather_humidity': None, 'weather_description': None})

        call_count += 1  # ✅ increment AFTER the request

    else:
        results.append({'index': i, 'weather_temp': None, 'weather_humidity': None, 'weather_description': None})

# Merge back to original df
weather_df = pd.DataFrame(results).set_index('index')
df = pd.concat([df, weather_df], axis=1)

print(f'Available data: {yes} / Total rows: {len(df)}')
```
